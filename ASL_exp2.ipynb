````python
import numpy as np

# Data
x = np.array([37388, 44758, 45833, 30862, 34212, 33190])
y = np.array([5318, 5061, 5008, 5795, 5283, 5259])
n = len(x)

# Regression coefficients
b = np.sum((x - np.mean(x))*(y - np.mean(y))) / np.sum((x - np.mean(x))**2)
a = np.mean(y) - b*np.mean(x)

# Predicted values
y_hat = a + b*x

# Residuals
residuals = y - y_hat

# Residual Sum of Squares
RSS = np.sum(residuals**2)

# Standard error of slope
SE_b = np.sqrt(RSS / (n-2) / np.sum((x - np.mean(x))**2))

# Standard error of intercept
SE_a = SE_b * np.sqrt(np.sum(x**2)/n)

# t-statistics
t_b = b / SE_b
t_a = a / SE_a

print(f"Slope (b) t-statistic: {t_b:.3f}")
print(f"Intercept (a) t-statistic: {t_a:.3f}")

# Standard errors
SE_b = np.sqrt(RSS / (n-2) / np.sum((x - np.mean(x))**2))
SE_a = SE_b * np.sqrt(np.sum(x**2)/n)

print(f"Standard Error of slope (SE_b): {SE_b:.4f}")
print(f"Standard Error of intercept (SE_a): {SE_a:.2f}")
````

````python
import numpy as np
from scipy import stats

# Data
x = np.array([37388, 44758, 45833, 30862, 34212, 33190])
y = np.array([5318, 5061, 5008, 5795, 5283, 5259])
n = len(x)

# Regression coefficients
b = np.sum((x - np.mean(x))*(y - np.mean(y))) / np.sum((x - np.mean(x))**2)
a = np.mean(y) - b*np.mean(x)

# Predicted values and residuals
y_hat = a + b*x
residuals = y - y_hat
RSS = np.sum(residuals**2)

# Standard errors
SE_b = np.sqrt(RSS / (n-2) / np.sum((x - np.mean(x))**2))
SE_a = SE_b * np.sqrt(np.sum(x**2)/n)

# t-statistics
t_b = b / SE_b
t_a = a / SE_a

# Two-tailed p-values
df = n - 2  # degrees of freedom
p_b = 2 * (1 - stats.t.cdf(np.abs(t_b), df))
p_a = 2 * (1 - stats.t.cdf(np.abs(t_a), df))

print(f"Slope (b) p-value: {p_b:.4f}")
print(f"Intercept (a) p-value: {p_a:.4f}")
````

````python
import statsmodels.api as sm
from scipy import stats

# Dependent variable
y = df['Sales']

# -------- Full Model (TV + intercept) --------
X_full = sm.add_constant(df['TV'])
full_model = sm.OLS(y, X_full).fit()

# -------- Restricted Model (Intercept only) --------
X_restricted = sm.add_constant(pd.DataFrame({'const': [1]*len(df)}))
restricted_model = sm.OLS(y, X_restricted).fit()

# -------- Likelihood Ratio Test --------
LR_stat = 2 * (full_model.llf - restricted_model.llf)

df_diff = full_model.df_model - restricted_model.df_model  # degrees of freedom difference

p_value = stats.chi2.sf(LR_stat, df_diff)

print("===== Likelihood Ratio Test =====")
print("LR Statistic:", LR_stat)
print("Degrees of Freedom:", int(df_diff))
print("P-value:", p_value)
````

````python
import statsmodels.api as sm
import pandas as pd
from scipy import stats

# Dependent variable
y = df['Sales']

# -------- Full Model (TV + intercept) --------
X_full = sm.add_constant(df['TV'])
full_model = sm.OLS(y, X_full).fit()

# -------- Restricted Model (Intercept only) --------
X_restricted = sm.add_constant(pd.DataFrame(index=df.index))
restricted_model = sm.OLS(y, X_restricted).fit()

# -------- Likelihood Ratio Statistic --------
LR_stat = 2 * (full_model.llf - restricted_model.llf)

# Degrees of freedom difference
df_diff = int(full_model.df_model - restricted_model.df_model)

# Critical value from Chi-square table (alpha = 0.05)
chi_critical = stats.chi2.ppf(0.95, df_diff)

print("===== Likelihood Ratio Test =====")
print("LR Statistic:", LR_stat)
print("Degrees of Freedom:", df_diff)
print("Chi-square Critical Value (0.05, df=1):", chi_critical)
# Decision based on Likelihood Ratio
if LR_stat > chi_critical:
    print("Decision: Reject H0 (TV is significant).")
    print("Conclusion: Full model (with TV) is better than restricted model (intercept only).")
else:
    print("Decision: Fail to Reject H0 (TV is not significant).")
    print("Conclusion: Restricted model (intercept only) is sufficient.")
````

````python
import statsmodels.api as sm
import pandas as pd

# Prepare data
y = df['Sales']

# Linear model: Sales ~ TV
X_linear = sm.add_constant(df['TV'])
linear_model = sm.OLS(y, X_linear).fit()
aic_linear = linear_model.aic

# Quadratic model: Sales ~ TV + TV^2
df['TV_sq'] = df['TV']**2
X_quad = sm.add_constant(df[['TV', 'TV_sq']])
quad_model = sm.OLS(y, X_quad).fit()
aic_quad = quad_model.aic

# Print results
print("===== Model AIC Comparison =====")
print(f"Linear Model AIC: {aic_linear:.2f}")
print(f"Quadratic Model AIC: {aic_quad:.2f}")

# Decision
if aic_quad < aic_linear:
    print("Conclusion: Quadratic model is better (lower AIC).")
else:
    print("Conclusion: Linear model is better (lower AIC).")
````

### Interpretation of Residuals vs. Fitted Values Plot

The residuals vs. fitted values plot is crucial for assessing key assumptions of a linear regression model:

1.  **Homoscedasticity (Constant Variance)**: We look for a random scatter of residuals around the zero line with no discernible pattern. If the spread of residuals increases or decreases with the fitted values (a funnel or cone shape), it indicates heteroscedasticity, violating this assumption. In our plot, the residuals appear to be somewhat randomly scattered around zero, but there might be a slight indication of increasing variance as fitted values increase. This suggests that the model's errors might not have constant variance, which is a potential violation of homoscedasticity.

2.  **Linearity**: The residuals should be randomly distributed around the zero line without any clear curve or pattern. A distinct pattern (e.g., U-shape or inverted U-shape) would suggest that the linear model might not be appropriate, and a non-linear relationship exists between the independent and dependent variables. In this plot, there isn't a strong, clear curve, which is good. However, the initial suggestion of a better fit with a quadratic model (from the AIC comparison in the previous step) could be subtly hinted at here if there's any curvature, even a slight one, in the residual pattern.

3.  **Independence of Errors**: While this plot doesn't directly confirm independence (time series plots or Durbin-Watson statistic are better for that), a random scatter generally supports this assumption. Any clear pattern would suggest dependent errors.

**Conclusion based on the plot**: The plot shows a relatively random scatter of residuals around the zero line. However, the residuals' spread might be slightly wider for higher fitted values, hinting at mild heteroscedasticity. Also, given the previous AIC comparison favoring a quadratic model, the perceived 'randomness' could mask a subtle non-linearity that the quadratic term was able to capture. For a more robust model, addressing the potential heteroscedasticity (e.g., using robust standard errors or transformations) or confirming the non-linear relationship with a quadratic term would be beneficial.
