### Binomial Probabilities using Bernoulli's Equation (Binomial PMF formula)

```python
import math
from scipy.stats import binom

n = 20      # number of items inspected
p = 0.5    # probability an item is defective
k = 2       # defect threshold

# Define custom binomial_pmf function using Bernoulli trial principles
def custom_binomial_pmf(n, k_val, p):
    # math.comb(n, k_val) calculates 'n choose k_val'
    return math.comb(n, k_val) * (p**k_val) * ((1-p)**(n-k_val))

print("--- Calculations using the Binomial PMF formula (derived from Bernoulli's equation) ---")

# 1. Probability of exactly k defects
prob_exact_k_custom = custom_binomial_pmf(n, k, p)
print(f"P(X = {k}) = {prob_exact_k_custom:.6f}")

# 2. Probability of at most k defects (X <= k) - summing individual PMF values
prob_at_most_k_custom = sum(custom_binomial_pmf(n, i, p) for i in range(k + 1))
print(f"P(X ≤ {k}) = {prob_at_most_k_custom:.6f}")

# 3. Probability of more than k defects (X > k) - using complement rule
prob_more_than_k_custom = 1 - prob_at_most_k_custom
print(f"P(X > {k}) = {prob_more_than_k_custom:.6f}")

print("\n--- For comparison, using scipy's built-in functions ---")

# 1. Probability of exactly k defects (using scipy's binom.pmf)
prob_exact_k_scipy = binom.pmf(k, n, p)
print(f"P(X = {k}) = {prob_exact_k_scipy:.6f}")

# 2. Probability of at most k defects (X <= k) (using scipy's binom.cdf)
prob_at_most_k_scipy = binom.cdf(k, n, p)
print(f"P(X ≤ {k}) = {prob_at_most_k_scipy:.6f}")

# 3. Probability of more than k defects (X > k) (using scipy's binom.sf - survival function or 1 - cdf)
prob_more_than_k_scipy = binom.sf(k, n, p) # sf(k,n,p) = P(X > k)
print(f"P(X > {k}) = {prob_more_than_k_scipy:.6f}")
```

### Comparing Bernoulli and Binomial Distributions

The plots above visually demonstrate the relationship between Bernoulli and Binomial distributions:

*   **Bernoulli Distribution (left plot):** This represents a single trial with two possible outcomes: success (1) or failure (0). In this example, with `p = 0.3`, the probability of success is 0.3, and the probability of failure is `1 - p = 0.7`. This is essentially a Binomial distribution with `n=1`.

*   **Binomial Distribution (right plot):** This models the number of successes in a fixed number of independent Bernoulli trials. Here, with `n=10` trials and `p=0.3` for each trial, the graph shows the probabilities of getting 0, 1, 2, ..., up to 10 successes. You can see how the probabilities are distributed across these possible counts of successes.

**Interpretation:**
*   The **Bernoulli distribution** is the foundational building block. It describes the probability of an outcome (success or failure) for *one* event.
*   The **Binomial distribution** is a generalization of the Bernoulli distribution. It describes the probability of obtaining a certain number of successes ($k$) in a given number of *independent and identical Bernoulli trials* ($n$). Each trial within the binomial process follows a Bernoulli distribution. The binomial distribution's shape changes with `n` and `p`, becoming more bell-shaped as `n` increases (approaching a normal distribution under certain conditions).
